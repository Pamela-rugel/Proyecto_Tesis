```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd()))
import pandas as pd
import _preprocesamiento_comun as pc

pd.set_option('display.max_columns', 100)
```

## 1. Carga de datos crudos

```python
df = pc.leer_csv('ponentestodos.csv', low_memory=False)
df.head()
```

## 2. Exploración inicial

```python
pc.resumen(df, 'ponentes_todos')
```

```python
df.dtypes
```

## 3. Limpieza

```python
df = pc.limpiar_strings(df)
df = pc.quitar_columnas_vacias(df, umbral=0.987)
df = pc.quitar_columnas_constantes(df, excluir=["REVISADOPARAESCALAFON"])
df = df.drop(columns=['ULTIMO_CAMBIO'], errors='ignore')
```

## 5. Tipado de fechas e identificadores

```python
df = pc.castear_fechas(df, ['FECHAINICIO', 'FECHAFIN', 'FECHAASCENSO', 'FECHACHECKESCALAFON', 'FECHASUBIDAARCHIVO', 'FECHAAPROBACION'])
df = pc.castear_enteros(df, ['IDCAPACITACION', 'IDPERSONA', 'IDPAIS', 'IDCIUDAD', 'IDINSTITUCION', 'IDUSUARIO', 'IDCANTON', 'IDTIPOCONOCIMIEN'])
df = pc.convertir_sn_a_binario(df, ['ENESCALAFON', 'CONSIDERACOMISANT', 'INGRESORRHH', 'REVISADOPARAESCALAFON', 'GESTIONUATHESPOL'])
```
