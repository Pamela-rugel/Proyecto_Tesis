```python
df = pc.castear_fechas(df, ['FECHAINICIO', 'FECHAFIN', 'ULTIMO_CAMBIO'])
df = pc.castear_enteros(df, ['IDPROYECTOINVESTIGACION', 'IDPERSONA', 'IDPAIS', 'IDROLPROYECTO', 'IDNOMENCLATURATIT', 'IDNOMENCLATURATIT2', 'IDAREACONOCIM', 'IDSUBAREACONOC', 'IDAREACONOCIM2', 'IDSUBAREACONOC2', 'IDAREAFRASCATI', 'IDSUBAREAFRASCATI', 'IDAREAFRASCATI2', 'IDSUBAREAFRASCATI2'])
df = pc.convertir_sn_a_binario(df, ['ENESCALAFON'])
df = df.drop(columns=['ULTIMO_CAMBIO'], errors='ignore')
```
